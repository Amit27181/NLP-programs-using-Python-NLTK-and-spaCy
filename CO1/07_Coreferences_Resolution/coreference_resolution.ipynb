```python
import spacy

nlp = spacy.load("en_core_web_sm")
```

```python
text = """
Rahul bought a new laptop.
He uses it every day.
The laptop helps Rahul with his studies.
"""

print(text)
```

```python
import re

text = """
Rahul bought a new laptop.
He uses it every day.
The laptop helps Rahul with his studies.
"""

# Simple example-specific pronoun replacement
resolved_text = re.sub(r"\bHe\b", "Rahul", text)
resolved_text = re.sub(r"\bit\b", "the laptop", resolved_text)

print("Original Text:\n")
print(text)

print("\nResolved Text:\n")
print(resolved_text)
```

```python
doc = nlp(text)

print("Named Entities:")
for ent in doc.ents:
    print(ent.text, "->", ent.label_)

print("\nPronouns:")
for token in doc:
    if token.pos_ == "PRON":
        print(token.text)
```

```python
print("Coreference Resolution Example")
print("-" * 35)
print("Original: Rahul bought a new laptop. He uses it every day.")
print("Resolved: Rahul bought a new laptop. Rahul uses the laptop every day.")
```
