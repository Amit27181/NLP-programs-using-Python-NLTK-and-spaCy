```python
import torch
from transformers import MarianMTModel, MarianTokenizer
```

```python
model_name = "Helsinki-NLP/opus-mt-en-hi"

tokenizer = MarianTokenizer.from_pretrained(model_name)
model = MarianMTModel.from_pretrained(model_name)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = model.to(device)

print("Model loaded successfully!")
print("Device:", device)
```

```python
def translate_text(text):
    inputs = tokenizer(
        text,
        return_tensors="pt",
        padding=True,
        truncation=True,
        max_length=128
    )

    inputs = {key: value.to(device) for key, value in inputs.items()}

    with torch.no_grad():
        translated = model.generate(
            **inputs,
            max_new_tokens=128,
            num_beams=4
        )

    return tokenizer.decode(
        translated[0],
        skip_special_tokens=True
    )
```

```python
text = "Natural Language Processing is a branch of Artificial Intelligence."

translation = translate_text(text)

print("English:", text)
print("Hindi:", translation)
```

```python
sentences = [
    "How are you?",
    "I love learning new things.",
    "Artificial Intelligence is changing the world.",
    "Machine translation converts text from one language to another.",
    "India is a beautiful country."
]

for sentence in sentences:
    print("English:", sentence)
    print("Hindi:", translate_text(sentence))
    print("-" * 60)
```

```python
while True:
    text = input("Enter an English sentence (or type 'exit'): ")

    if text.strip().lower() == "exit":
        print("Translation completed.")
        break

    if not text.strip():
        print("Please enter a sentence.")
        continue

    print("Hindi Translation:", translate_text(text))
```
