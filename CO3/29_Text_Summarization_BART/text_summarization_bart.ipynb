```python
import torch
from transformers import BartTokenizer, BartForConditionalGeneration
```

```python
model_name = "facebook/bart-large-cnn"

tokenizer = BartTokenizer.from_pretrained(model_name)
model = BartForConditionalGeneration.from_pretrained(model_name)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = model.to(device)

print("Device:", device)
print("BART model loaded successfully!")
```

```python
def summarize_text(text, max_length=100, min_length=30):
    inputs = tokenizer(
        text,
        return_tensors="pt",
        max_length=1024,
        truncation=True
    ).to(device)

    with torch.no_grad():
        summary_ids = model.generate(
            inputs["input_ids"],
            attention_mask=inputs["attention_mask"],
            max_length=max_length,
            min_length=min_length,
            num_beams=4,
            early_stopping=True
        )

    summary = tokenizer.decode(
        summary_ids[0],
        skip_special_tokens=True
    )

    return summary
```

```python
text = """
Artificial Intelligence is transforming many industries, including
healthcare, education, agriculture, and transportation. In healthcare,
AI helps doctors analyze medical images, predict diseases, and support
clinical decisions. In education, AI-powered tools provide personalized
learning experiences. Agriculture uses AI for crop monitoring, disease
detection, and yield prediction. Although AI offers many benefits,
challenges such as data privacy, bias, and ethical concerns must be
addressed to ensure responsible use.
"""

summary = summarize_text(text)

print("Original Text:\n")
print(text)

print("\nGenerated Summary:\n")
print(summary)
```

```python
original_words = len(text.split())
summary_words = len(summary.split())

print("Original word count:", original_words)
print("Summary word count:", summary_words)

if original_words > 0:
    reduction = (1 - summary_words / original_words) * 100
    print(f"Text length reduction: {reduction:.2f}%")
```

```python
user_text = input("Enter a paragraph to summarize:\n")

if len(user_text.strip()) < 30:
    print("Please enter a longer paragraph for meaningful summarization.")
else:
    result = summarize_text(user_text)
    print("\nGenerated Summary:\n")
    print(result)
```
