```python
import nltk
import gensim
from gensim.models import Word2Vec
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords

print("NLTK version:", nltk.__version__)
print("Gensim version:", gensim.__version__)
```

```python
nltk.download("punkt")
nltk.download("punkt_tab")
nltk.download("stopwords")
```

```python
corpus = [
    "Natural language processing helps computers understand human language",
    "Machine learning is a part of artificial intelligence",
    "Deep learning uses neural networks for learning",
    "Word embeddings represent words as numerical vectors",
    "Word2Vec learns word relationships from text data",
    "Natural language processing uses machine learning techniques",
    "Artificial intelligence includes machine learning and deep learning",
    "Neural networks are useful for natural language processing",
    "Word embeddings help computers process textual information",
    "Machine learning models learn patterns from data"
]

print("Total sentences:", len(corpus))
for sentence in corpus:
    print("-", sentence)
```

```python
stop_words = set(stopwords.words("english"))

tokenized_corpus = []

for sentence in corpus:
    tokens = word_tokenize(sentence.lower())
    tokens = [
        word for word in tokens
        if word.isalpha() and word not in stop_words
    ]
    tokenized_corpus.append(tokens)

print("Preprocessed corpus:")
for tokens in tokenized_corpus:
    print(tokens)
```

```python
model = Word2Vec(
    sentences=tokenized_corpus,
    vector_size=100,
    window=3,
    min_count=1,
    workers=1,
    sg=1,
    seed=42,
    epochs=100
)

print("Word2Vec model trained successfully!")
print("Vocabulary size:", len(model.wv))
```

```python
print("Vocabulary:")
print(list(model.wv.index_to_key))
```

```python
word = "language"

if word in model.wv:
    vector = model.wv[word]
    print("Word:", word)
    print("Vector dimensions:", len(vector))
    print("Word vector:\n", vector)
else:
    print("Word not found in vocabulary.")
```

```python
word = "learning"

if word in model.wv:
    print("Words similar to:", word)
    for similar_word, score in model.wv.most_similar(word, topn=5):
        print(f"{similar_word}: {score:.4f}")
else:
    print("Word not found in vocabulary.")
```

```python
word1 = "language"
word2 = "learning"

if word1 in model.wv and word2 in model.wv:
    similarity = model.wv.similarity(word1, word2)
    print(f"Similarity between '{word1}' and '{word2}': {similarity:.4f}")
else:
    print("One or both words are missing from vocabulary.")
```

```python
model.save("word2vec_model.model")
print("Model saved successfully!")

loaded_model = Word2Vec.load("word2vec_model.model")
print("Model loaded successfully!")
print("Vocabulary size:", len(loaded_model.wv))
```
