```python
import re
import numpy as np
from gensim.models import Word2Vec

print("Libraries imported successfully!")
```

```python
documents = [
    "the cat sits on the mat",
    "the kitten sits on the rug",
    "the dog plays in the garden",
    "a puppy plays in the park"
]

for i, document in enumerate(documents, start=1):
    print(f"Document {i}: {document}")
```

```python
def tokenize(text):
    return re.findall(r"\b[a-z]+\b", text.lower())

tokenized_documents = [tokenize(doc) for doc in documents]

for tokens in tokenized_documents:
    print(tokens)
```

```python
model = Word2Vec(
    sentences=tokenized_documents,
    vector_size=50,
    window=3,
    min_count=1,
    workers=1,
    seed=42,
    epochs=200
)

print("Word embeddings trained successfully!")
print("Vocabulary size:", len(model.wv))
```

```python
pairs = [
    (0, 1),
    (0, 2),
    (2, 3)
]

for i, j in pairs:
    distance = model.wv.wmdistance(
        tokenized_documents[i],
        tokenized_documents[j]
    )

    print(f"\nDocument {i + 1}: {documents[i]}")
    print(f"Document {j + 1}: {documents[j]}")
    print(f"WMD: {distance:.4f}")
```

```python
results = []

for i in range(len(documents)):
    for j in range(i + 1, len(documents)):
        distance = model.wv.wmdistance(
            tokenized_documents[i],
            tokenized_documents[j]
        )
        results.append((i, j, distance))

results.sort(key=lambda item: item[2])

i, j, distance = results[0]

print("Most similar document pair:")
print("Document 1:", documents[i])
print("Document 2:", documents[j])
print(f"Word Mover's Distance: {distance:.4f}")
```
