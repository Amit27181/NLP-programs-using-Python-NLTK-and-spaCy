```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.util import ngrams

nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)

print("N-Gram setup completed!")
```

```python
from sklearn.feature_extraction.text import CountVectorizer

# Custom corpus
documents = [
    "I love natural language processing",
    "I love machine learning",
    "Natural language processing is interesting"
]

# ---------------- UNIGRAM ----------------
unigram_vectorizer = CountVectorizer(ngram_range=(1, 1))
unigram_matrix = unigram_vectorizer.fit_transform(documents)

print("===== UNIGRAM =====")
print("\nVocabulary:")
print(unigram_vectorizer.get_feature_names_out())

print("\nRepresentation:")
print(unigram_matrix.toarray())


# ---------------- BIGRAM ----------------
bigram_vectorizer = CountVectorizer(ngram_range=(2, 2))
bigram_matrix = bigram_vectorizer.fit_transform(documents)

print("\n===== BIGRAM =====")
print("\nVocabulary:")
print(bigram_vectorizer.get_feature_names_out())

print("\nRepresentation:")
print(bigram_matrix.toarray())


# ---------------- TRIGRAM ----------------
trigram_vectorizer = CountVectorizer(ngram_range=(3, 3))
trigram_matrix = trigram_vectorizer.fit_transform(documents)

print("\n===== TRIGRAM =====")
print("\nVocabulary:")
print(trigram_vectorizer.get_feature_names_out())

print("\nRepresentation:")
print(trigram_matrix.toarray())
```
