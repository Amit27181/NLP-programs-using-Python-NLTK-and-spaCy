```python
import numpy as np
from gensim.models import KeyedVectors

print("Libraries imported successfully!")
```

```python
glove_vectors = {
    "king": np.array([0.8, 0.6, 0.1, 0.9]),
    "queen": np.array([0.7, 0.7, 0.2, 0.8]),
    "man": np.array([0.6, 0.3, 0.1, 0.5]),
    "woman": np.array([0.5, 0.4, 0.2, 0.6]),
    "apple": np.array([0.1, 0.2, 0.9, 0.3]),
    "orange": np.array([0.2, 0.3, 0.8, 0.4])
}

print("Sample embeddings created successfully!")
```

```python
word = "king"

if word in glove_vectors:
    print("Word:", word)
    print("Vector:", glove_vectors[word])
    print("Dimensions:", len(glove_vectors[word]))
else:
    print("Word not found.")
```

```python
def cosine_similarity(vec1, vec2):
    denominator = np.linalg.norm(vec1) * np.linalg.norm(vec2)

    if denominator == 0:
        return 0.0

    return float(np.dot(vec1, vec2) / denominator)


word1 = "king"
word2 = "queen"

similarity = cosine_similarity(
    glove_vectors[word1],
    glove_vectors[word2]
)

print(f"Cosine similarity ({word1}, {word2}): {similarity:.4f}")
```

```python
word_pairs = [
    ("king", "queen"),
    ("man", "woman"),
    ("apple", "orange"),
    ("king", "apple")
]

for word1, word2 in word_pairs:
    score = cosine_similarity(
        glove_vectors[word1],
        glove_vectors[word2]
    )
    print(f"{word1} - {word2}: {score:.4f}")
```

```python
import numpy as np

glove_vectors = {
    "king":   np.array([0.8, 0.6, 0.1, 0.9]),
    "queen":  np.array([0.7, 0.7, 0.2, 0.8]),
    "man":    np.array([0.6, 0.3, 0.1, 0.5]),
    "woman":  np.array([0.5, 0.4, 0.2, 0.6]),
    "apple":  np.array([0.1, 0.2, 0.9, 0.3]),
    "orange": np.array([0.2, 0.3, 0.8, 0.4])
}

print("GloVe-style sample embeddings created successfully!")
print("Vocabulary size:", len(glove_vectors))
print("Vector dimensions:", len(glove_vectors["king"]))
```

```python
def cosine_similarity(vec1, vec2):
    denominator = np.linalg.norm(vec1) * np.linalg.norm(vec2)
    if denominator == 0:
        return 0.0
    return float(np.dot(vec1, vec2) / denominator)


word = "king"

print("Word:", word)
print("Vector:", glove_vectors[word])

scores = []

for other_word, vector in glove_vectors.items():
    if other_word != word:
        score = cosine_similarity(glove_vectors[word], vector)
        scores.append((other_word, score))

scores.sort(key=lambda item: item[1], reverse=True)

print("\nMost similar sample words:")
for other_word, score in scores[:3]:
    print(f"{other_word}: {score:.4f}")
```
