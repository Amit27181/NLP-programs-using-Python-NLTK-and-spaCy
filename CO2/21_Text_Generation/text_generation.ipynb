```python
import random
import re
from collections import defaultdict, Counter
```

```python
corpus = """
Natural language processing enables computers to understand human language.
Natural language processing is used in chatbots and virtual assistants.
Machine learning helps computers learn patterns from data.
Deep learning uses neural networks to learn complex patterns.
Text generation is an important application of natural language processing.
Language models learn the relationships between words in a sentence.
Artificial intelligence is transforming education healthcare and business.
Python is widely used for machine learning and natural language processing.
A language model can generate text by predicting the next word.
Text generation can be used for stories summaries and chatbot responses.
"""

print("Sample Corpus:\n")
print(corpus)
```

```python
def preprocess_text(text):
    text = text.lower()
    text = re.sub(r"[^a-z\s]", "", text)
    return text.split()

tokens = preprocess_text(corpus)

print("Total words:", len(tokens))
print("First 30 tokens:", tokens[:30])
```

```python
bigram_model = defaultdict(Counter)

for i in range(len(tokens) - 1):
    current_word = tokens[i]
    next_word = tokens[i + 1]
    bigram_model[current_word][next_word] += 1

print("Example next-word probabilities for 'natural':")
print(bigram_model["natural"])
```

```python
def get_next_word_probabilities(word):
    word = word.lower()

    if word not in bigram_model:
        return {}

    counts = bigram_model[word]
    total = sum(counts.values())

    return {
        next_word: count / total
        for next_word, count in counts.items()
    }

print(get_next_word_probabilities("natural"))
```

```python
def generate_text(start_word, num_words=20):
    current_word = start_word.lower()
    generated_words = [current_word]

    for _ in range(num_words - 1):
        next_words = bigram_model.get(current_word)

        if not next_words:
            break

        candidates = list(next_words.keys())
        weights = list(next_words.values())

        current_word = random.choices(
            candidates,
            weights=weights,
            k=1
        )[0]

        generated_words.append(current_word)

    return " ".join(generated_words)

print(generate_text("natural", 20))
```

```python
start_words = ["natural", "machine", "text", "language", "python"]

for word in start_words:
    print(f"\nStarting word: {word}")
    print(generate_text(word, 15))
```

```python
random_start = random.choice(tokens)

print("Random starting word:", random_start)
print("\nGenerated text:")
print(generate_text(random_start, 25))
```

```python
user_word = input("Enter a starting word: ").strip().lower()

if user_word in bigram_model:
    number_of_words = int(input("How many words should be generated? "))

    if number_of_words < 1:
        print("Enter a number greater than zero.")
    else:
        print("\nGenerated Text:")
        print(generate_text(user_word, min(number_of_words, 100)))
else:
    print("Word not found in the language model. Try another word.")
```

```python
trigram_model = defaultdict(Counter)

for i in range(len(tokens) - 2):
    context = (tokens[i], tokens[i + 1])
    next_word = tokens[i + 2]
    trigram_model[context][next_word] += 1

print("Example trigram context:")
print(("natural", "language"))
print("Possible next words:", trigram_model[("natural", "language")])
```

```python
def generate_trigram_text(word1, word2, num_words=20):
    word1 = word1.lower()
    word2 = word2.lower()

    generated_words = [word1, word2]

    for _ in range(num_words - 2):
        context = (generated_words[-2], generated_words[-1])
        next_words = trigram_model.get(context)

        if not next_words:
            break

        candidates = list(next_words.keys())
        weights = list(next_words.values())

        next_word = random.choices(
            candidates,
            weights=weights,
            k=1
        )[0]

        generated_words.append(next_word)

    return " ".join(generated_words[:num_words])

print(generate_trigram_text("natural", "language", 20))
```

```python
print("BIGRAM MODEL:")
print(generate_text("natural", 15))

print("\nTRIGRAM MODEL:")
print(generate_trigram_text("natural", "language", 15))
```

```python
generated_output = generate_text("machine", 25)

with open("generated_text.txt", "w", encoding="utf-8") as file:
    file.write(generated_output)

print("Generated text saved to generated_text.txt")
print(generated_output)
```
