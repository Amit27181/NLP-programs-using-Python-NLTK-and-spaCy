```python
import re
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.svm import LinearSVC
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

print("Libraries imported successfully!")
```

```python
texts = [
    "I love this product",
    "This is an amazing experience",
    "The service is excellent",
    "I am very happy with this purchase",
    "This product works perfectly",
    "Absolutely wonderful experience",
    "I hate this product",
    "This is a terrible experience",
    "The service is very poor",
    "I am disappointed with this purchase",
    "This product does not work",
    "Absolutely horrible experience",
    "The quality is fantastic",
    "I really like this service",
    "The product is very useful",
    "The quality is awful",
    "I dislike this service",
    "This product is useless",
    "I am satisfied with the result",
    "The experience was disappointing"
]

labels = [
    "Positive", "Positive", "Positive", "Positive", "Positive",
    "Positive", "Negative", "Negative", "Negative", "Negative",
    "Negative", "Negative", "Positive", "Positive", "Positive",
    "Negative", "Negative", "Negative", "Positive", "Negative"
]

print("Total documents:", len(texts))
print("Total labels:", len(labels))
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    texts,
    labels,
    test_size=0.2,
    random_state=42,
    stratify=labels
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

```python
vectorizer = TfidfVectorizer(
    lowercase=True,
    stop_words="english",
    ngram_range=(1, 2)
)

X_train_tfidf = vectorizer.fit_transform(X_train)
X_test_tfidf = vectorizer.transform(X_test)

print("Training matrix shape:", X_train_tfidf.shape)
print("Testing matrix shape:", X_test_tfidf.shape)
```

```python
nb_model = MultinomialNB()
nb_model.fit(X_train_tfidf, y_train)

nb_predictions = nb_model.predict(X_test_tfidf)

print("Naive Bayes model trained successfully!")
print("Accuracy:", round(accuracy_score(y_test, nb_predictions) * 100, 2), "%")
```

```python
print("Naive Bayes Classification Report:\n")
print(classification_report(
    y_test,
    nb_predictions,
    labels=["Negative", "Positive"],
    zero_division=0
))

print("Confusion Matrix:")
print(confusion_matrix(
    y_test,
    nb_predictions,
    labels=["Negative", "Positive"]
))
```

```python
svm_model = LinearSVC(random_state=42)
svm_model.fit(X_train_tfidf, y_train)

svm_predictions = svm_model.predict(X_test_tfidf)

print("SVM model trained successfully!")
print("Accuracy:", round(accuracy_score(y_test, svm_predictions) * 100, 2), "%")
```

```python
nb_accuracy = accuracy_score(y_test, nb_predictions)
svm_accuracy = accuracy_score(y_test, svm_predictions)

print("Model Performance Comparison")
print("-" * 35)
print(f"Naive Bayes Accuracy: {nb_accuracy * 100:.2f}%")
print(f"SVM Accuracy:         {svm_accuracy * 100:.2f}%")

if nb_accuracy > svm_accuracy:
    print("Better test accuracy: Naive Bayes")
elif svm_accuracy > nb_accuracy:
    print("Better test accuracy: SVM")
else:
    print("Both models have equal test accuracy.")
```

```python
new_texts = [
    "I really love this amazing product",
    "This is a horrible and useless product"
]

new_vectors = vectorizer.transform(new_texts)

nb_results = nb_model.predict(new_vectors)
svm_results = svm_model.predict(new_vectors)

for text, nb_result, svm_result in zip(
    new_texts, nb_results, svm_results
):
    print("\nText:", text)
    print("Naive Bayes Prediction:", nb_result)
    print("SVM Prediction:", svm_result)
```
