```python
from textblob import TextBlob
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer

analyzer = SentimentIntensityAnalyzer()

print("Libraries imported successfully!")
```

```python
texts = [
    "I absolutely love this product. It is amazing!",
    "The movie was terrible and boring.",
    "The product arrived today.",
    "The food was delicious and the service was excellent.",
    "I am very disappointed with this purchase.",
    "The weather is normal today."
]

for i, text in enumerate(texts, start=1):
    print(f"{i}. {text}")
```

```python
def get_textblob_sentiment(text):
    polarity = TextBlob(text).sentiment.polarity

    if polarity > 0:
        sentiment = "Positive"
    elif polarity < 0:
        sentiment = "Negative"
    else:
        sentiment = "Neutral"

    return polarity, sentiment


print("TextBlob Sentiment Analysis\n")

for text in texts:
    polarity, sentiment = get_textblob_sentiment(text)
    print("Text:", text)
    print(f"Polarity: {polarity:.3f} | Sentiment: {sentiment}\n")
```

```python
print("VADER Sentiment Analysis\n")

for text in texts:
    scores = analyzer.polarity_scores(text)
    compound = scores["compound"]

    if compound >= 0.05:
        sentiment = "Positive"
    elif compound <= -0.05:
        sentiment = "Negative"
    else:
        sentiment = "Neutral"

    print("Text:", text)
    print("Scores:", scores)
    print("Sentiment:", sentiment)
    print()
```

```python
print(f"{'Text':55} {'TextBlob':12} {'VADER'}")
print("-" * 85)

for text in texts:
    _, tb_sentiment = get_textblob_sentiment(text)

    vader_compound = analyzer.polarity_scores(text)["compound"]

    if vader_compound >= 0.05:
        vader_sentiment = "Positive"
    elif vader_compound <= -0.05:
        vader_sentiment = "Negative"
    else:
        vader_sentiment = "Neutral"

    print(f"{text[:53]:55} {tb_sentiment:12} {vader_sentiment}")
```

```python
user_text = input("Enter a sentence for sentiment analysis: ")

tb_polarity, tb_sentiment = get_textblob_sentiment(user_text)

vader_scores = analyzer.polarity_scores(user_text)
compound = vader_scores["compound"]

if compound >= 0.05:
    vader_sentiment = "Positive"
elif compound <= -0.05:
    vader_sentiment = "Negative"
else:
    vader_sentiment = "Neutral"

print("\nTextBlob Result")
print("Polarity:", round(tb_polarity, 3))
print("Sentiment:", tb_sentiment)

print("\nVADER Result")
print("Scores:", vader_scores)
print("Sentiment:", vader_sentiment)
```

```python
from collections import Counter

textblob_results = [
    get_textblob_sentiment(text)[1]
    for text in texts
]

vader_results = []

for text in texts:
    compound = analyzer.polarity_scores(text)["compound"]

    if compound >= 0.05:
        vader_results.append("Positive")
    elif compound <= -0.05:
        vader_results.append("Negative")
    else:
        vader_results.append("Neutral")

print("TextBlob Summary:", dict(Counter(textblob_results)))
print("VADER Summary:", dict(Counter(vader_results)))
```
